```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import constants as cst
```

```python
def vt(p,q,s78,s89,s910):
    numerator   = p*q*s78*s89*s910
    denominator = p*q*s89*s910+s78*(p*q*s910 + s89*(p*q+(p+q)*s910))
    return numerator / denominator 


def vr(p,q,s78,s89,s910):
    numerator   = p*q*s89*s910+s78*(p*q*s910 + s89*(p*q+(p+q-p*q)*s910))
    denominator = p*q*s89*s910+s78*(p*q*s910 + s89*(p*q+(p+q)*s910))
    return numerator / denominator 
```

# Sanity check 

```python
from fractions import Fraction
x = np.array([1,3/4,2/4,1/4,1e-18])
y = np.array([1,3/4,2/4,1/4,1e-18])
X, Y = np.meshgrid(x, y)

fig, ax = plt.subplots(figsize = (6,5), layout = 'tight')

im = ax.pcolor(X,Y, vt(X,Y,1,1,1),cmap = 'magma')
ax.set_xticks([1,3/4,2/4,1/4,1e-18],
           ["1", "3/4", "2/4", "1/4","0"]);
ax.set_yticks([1,3/4,2/4,1/4,1e-18],
           ["1", "3/4", "2/4", "1/4","0"]);
cbar = plt.colorbar(im, ax=ax, label=r'$V_3/V_2$')
cbar.set_label(label=r'$V_3/V_2$', fontsize=15)

ax.set_xlabel(r'$p$', fontsize = 15)
ax.set_ylabel(r'$q$', fontsize = 15)

plt.savefig('sanity_check_1.png', dpi = 500, bbox_inches = 'tight')
```
